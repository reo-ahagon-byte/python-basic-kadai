```python
class Human:
  def __init__(self, name, age):
    # 属性を定義する
    self.name = name
    self.age = age

  def check_adult(self):
      if self.age >= 20:
       print(f"{self.name}は大人です")
      else:
        print(f"{self.name}大人ではありません")


human_list = []

taro = Human("太郎", 30)
hanako = Human("花子", 18)
ichiro = Human("一郎", 21)

human_list.append(taro)
human_list.append(hanako)
human_list.append(ichiro)

for person in human_list:
  person.check_adult()







```
